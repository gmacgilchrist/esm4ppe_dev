# Calculate Pearson correlations between two variables

```python
import esm4ppe
import xarray as xr
from dask.diagnostics import ProgressBar
import os
```

```python
from functions import *
```

```python
variable1 = 'intpppico'
variable2 = 'limirrpico'
frequency = 'monthly'
bymonth=True
detrend=False
deseasonalize=False
overwrite = False
```

```python
path = get_path_pearson(variable1,
                        variable2,
                        bymonth=bymonth,
                        detrend=detrend,
                        deseasonalize=deseasonalize)    
```

```python
if (os.path.exists(path)) & (overwrite==False):
    print(path+" exists.")
else:
    # load data
    ds = xr.Dataset()
    for variable in [variable1,variable2]:
        es = esm4ppe.esm4ppeObj(variable,frequency)
        ds[variable] = es.add_control().control[variable]
    static = es.static
    # calculate correlation
    cr,savefilelist = calc_pearson(ds[variable1],
                                   ds[variable2],
                                   bymonth=bymonth,
                                   detrend=detrend,
                                   deseasonalize=deseasonalize)
    # compute
    print("saving to "+path)
    with ProgressBar():
        cr = cr.compute()
    cr.to_netcdf(path,mode='w')
```
